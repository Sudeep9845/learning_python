### Inheritance
- It is a mechanism where a new class inherits properties and behaviors (methods and attributes) from an existing class.
- It represents an "Is-A" relationship (e.g., Student is a User).
- Parent Class (Base/Super Class): The class being inherited from.
- Child Class (Derived/Sub Class): The class that inherits.
- Benefits: Code Reusability and cleaner code structure.
- In this example: 'User' is the parent class; 'Tutor' and 'Student' are child classes.

## Single Inheritance
This is the simplest form. A single Child class inherits from a single Parent class.

- **Structure:** `Parent -> Child`
- **Use Case:** Extending basic functionality (e.g., A `Car` is a `Vehicle`).

```python
class Vehical:
    def start(self):
        print("The vehicle is starting.")
class Car(Vehical):
    def start(self):
        print("The car is starting with a roar!")
car1 = Car()
car1.start()  # Output: The car is starting with a roar!
```

## Multiple Inhritance

A single Child class inherits from **two or more** Parent classes at the same time.

- **Structure:** `Parent A + Parent B -> Child`
- **Uniqueness:** Python is one of the few languages (unlike Java/C#) that supports this natively.
- **Use Case:** A `FlyingCar` inherits features from both `Car` and `Airplane`.

> Multiple inheritance in Python follows MRO (Method Resolution Order)

```python
class Father:
    def show(self):
        print("Hello from Father")
class Mother:
    def show(self):
        print("Hello from Mother")
class Child(Father, Mother):
    def show_child(self):
        print("Hello from Child")
child1 = Child()
# Demonstrating Method Resolution Order (MRO)
child1.show()   # Output: Hello from Father (as Father is the first parent class)
child1.show_child()   # Output: Hello from Child
```

## Multilevel Inheritance
A class inherits from a Parent, which in turn inherits from another Parent (Grandparent). It forms a vertical chain.

- **Structure:** `Grandparent -> Parent -> Child`
- **Use Case:** `Puppy` inherits from `Dog`, which inherits from `Animal`.

```python
class Grandparent:
    def height(self):
        print("Tall height")

class Parent(Grandparent):
    def wisdom(self):
        print("Medium wisdom")

class Child(Parent):
    def energy(self):
        print("High energy")

c = Child()
c.height() # Inherited from Grandparent
c.wisdom() # Inherited from Parent
c.energy() # Own
```

## Hierarchical Inheritance

Multiple different Child classes inherit from the **same** Parent class.

- **Structure:** `Parent -> Child A, Child B, Child C`
- **Use Case:** `Cat`, `Dog`, and `Lion` all inherit from `Animal`.

```python
class Vehicle:
    def info(self):
        print("I am a Vehicle")

class Car(Vehicle):
    def drive(self):
        print("Car drives on roads")

class Boat(Vehicle):
    def sail(self):
        print("Boat sails on water")

c = Car()
b = Boat()

c.info() # Shared Parent method
b.info() # Shared Parent method
```

## Hybrid Inheritance

This is a combination of more than one type of inheritance mentioned above. It usually involves a mix of hierarchical and multiple inheritance.

- **Structure:** Complex mix (e.g., A Diamond shape).
- **Use Case:** A complex system where `University` inherits from `Institution`, and `Student` inherits from both `Person` and `University`.

```python
class A:
    pass

class B(A):
    pass
class C(A):
    pass
class D(B,C):
    pass

obj1 = D()
print(isinstance(obj1, D))  # True
```
