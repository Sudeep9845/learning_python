### Polymorphism
In programming, Polymorphism is the ability of a single interface (method or operator) to support objects of multiple different types. It allows you to treat different objects in a uniform way, even if they behave differently behind the scenes.

## Method Overriding
**"New Logic for the Same Name"**

- **Definition:** Happens when a Child class provides a specific implementation of a method that is already provided by its Parent class.
- **Context:** Requires **Inheritance** (Parent -> Child).
- **Signature:** The method name and parameters (usually) remain the **same**.
- **Purpose:** To change the behavior of an inherited method.

```python
class Animal:
    def speak(self):
        print("The animal makes a sound.")

class Dog(Animal):
    def speak(self):
        print("The dog barks")

dog1 = Dog()
dog1.speak()  # Output: The dog barks
```

## Method Overloading
**"Same Name, Different Arguments"**

- **Definition:** Creating multiple methods with the **same name** but different **parameters** (different number or types of arguments) within the same class.
- **Context:** Happens inside a **single class**.
- **Purpose:** To allow a function to handle different types of inputs cleanly.

> Python does NOT support traditional Method Overloading. But we can use "if else" as a work round

```python
class Calculator:
    def add(self,a,b,c=None):
        if c is not None:
            return a + b + c
        else:
            return a + b

calc = Calculator()
print(calc.add(2, 3))        # Output: 5
print(calc.add(2, 3, 4))     # Output: 9
```

## Operator Overloading
**Operator Overloading** allows you to define how standard Python operators (like `+`, `-`, `*`, `==`, `>`, etc.) behave when applied to your custom objects.

By default, Python knows how to add numbers (`5 + 10`) and concatenate strings (`"Hello" + "World"`). However, if you create a custom class like `Point` or `Employee`, Python doesn't know what `+` means for them. Operator overloading lets you teach Python what to do.

- **The Mechanism:** Python uses special methods (often called **Magic Methods** or **Dunder Methods**, short for "Double Underscore") to handle operators.
- **The Logic:** When you write `a + b`, Python automatically looks for a method named `__add__` inside object `a` and passes object `b` to it.

```python
from math import gcd
class Fraction:
    def __init__(self,numerator,denominator):
        self.numerator = numerator
        self.denominator = denominator

    def __add__(self,other):
        new_numerator = (self.numerator * other.denominator) + (other.numerator * self.denominator)
        new_denominator = self.denominator * other.denominator
        return Fraction(new_numerator, new_denominator).simplify()
    
    def __str__(self):
        return f"{self.numerator}/{self.denominator}"
    
    def simplify(self):
        common_divisor = gcd(self.numerator, self.denominator)
        return Fraction(self.numerator // common_divisor, self.denominator // common_divisor)

f1 = Fraction(8,10)
f2 = Fraction(12,6)
result = f1 + f2
print(f"{f1} + {f2} = {result}") # Output: 8/10 + 12/6 = 14/5
```
