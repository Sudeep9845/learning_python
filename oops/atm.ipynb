## ATM Example

### Reference Variable

A Reference Variable is a variable that stores the memory address of an object, rather than the data itself.

### Encapsulation

- Nothing is truly hidden in Python
- To hide an attribute or method from others we can use "__"(double underscores)
- example: self.__pin
- Internally: _ATM__pin 

```python
class ATM:

    # Constructor to initialize balance and pin
    # __init__ is a special method in Python classes. 
    # It is called when an object is created from the class and allows the class to initialize the attributes of the class.
    def __init__(self):
        self.balance=0
        self.pin=""
        self.menu()

    def menu(self):
        user_input = input("Welcome to the ATM. Choose an option: 1. Create PIN 2. Deposit 3. Withdraw 4. Check Balance 5. Exit\n")

        if user_input == "1":
            self.create_pin()
        elif user_input == "2":
            self.deposit()
        elif user_input == "3":
            self.withdraw()
        elif user_input == "4":
            self.check_balance()
        elif user_input == "5":
            self.exit()
        else:
            print("Invalid option. Please try again.")
    def create_pin(self):
        self.pin = input("Enter a new PIN: ")
        print("PIN created successfully.")
    
    def deposit(self):
        temp = input("Enter your PIN: ")
        if temp == self.pin:
            amount = int(input("Enter amount to deposit: "))
            self.balance = self.balance + amount
            print(f"Deposited: {amount}. New balance: {self.balance}")
        else:
            print("Incorrect PIN.")
    
    def withdraw(self):
        temp = input("Enter your PIN: ")
        if temp == self.pin:
            amount = float(input("Enter amount to withdraw: "))
            if amount <= self.balance:
                self.balance = self.balance - amount
                print(f"Withdrew: {amount}. New balance: {self.balance}")
            else:
                print("Insufficient balance.")
        else:
            print("Incorrect PIN.")
    
    def check_balance(self):
        temp = input("Enter your PIN: ")
        if temp == self.pin:
            print(f"Current balance: {self.balance}")
        else:
            print("Incorrect PIN.")
    
    def exit(self):
        print("Thank you for using the ATM. Goodbye!")
        
        
```

```python
## main code
if __name__ == "__main__":
        # atm is a Reference variable that refers to the object of the ATM class.
        atm = ATM()
        atm.menu()
        
    
    
```
