## Creating a class and object

```python
class Car:
    def __init__(self):
        # attributes
        self.color = "red"
        self.brand = "Toyota"
        self.start_engine()
        self.stop_engine()
        
    # methods 
    def start_engine(self):
        print("Engine started")
    
    def stop_engine(self):
        print("Engine stopped")
    
```

```python
if __name__ == "__main__":
    # create an object of the Car class
    my_car = Car()
    print(f"Car color: {my_car.color}, Brand: {my_car.brand}")
```
