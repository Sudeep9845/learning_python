## Numpy Tricks and useful methods

```python
import numpy as np
```

### Sort - np.sort()

```python
# vector sorting
a = np.random.randint(1,50,20)
a
```

```python
np.sort(a)
```

```python
# matrix sorting
b = np.random.randint(1,50,20).reshape(5,4)
b
```

```python
# sort by rows
np.sort(b,axis=1)
```

```python
# sort by columns
np.sort(b,axis=0)
```

```python
# desending order
np.sort(b,axis=0)[::-1]
```

### Append - np.append()

- numpy appends values along side the mentioned axis at the end of the array

```python
np.append(a,200)
```

```python
np.append(b,np.ones((b.shape[0],1)),axis=1)
```

### Concatenate - np.concatenate()

- numpy.concatenate() function concatenate a sequence of arrays along an existing axis.

```python
a = np.arange(6).reshape(2,3)
b = np.arange(6,12).reshape(2,3)
a,b
```

```python
np.concatenate((a,b),axis=0) # vertical stacking
```

### Unique - np.unique()

- With the help of np.unique() method, we can get the unique values from an array given as parameter in np.unique() method.

```python
a = np.array([1,1,1,1,2,3,3,3,3,2,4,6,6])
np.unique(a)
```

### Expand dimension - np.expand_dims()

- With the help of Numpy.expand_dims() method, we can get the expanded dimensions of an array
  

```python
a = np.arange(12).reshape(3,4)
b = np.arange(12,24)
a,b
```

```python
a.shape,b.shape
```

```python
# expanding a's dimensions
np.expand_dims(a,axis=0)
```

```python
np.expand_dims(a,axis=0).shape
```

### cumsum - np.cusum()

- numpy.cumsum() function is used when we want to compute the cumulative sum of array elements over a given axis.

```python
a = np.random.randint(1,100,20)
a
```

```python
np.cumsum(a)
```

```python
b = a.reshape(4,5)
b
```

```python
np.cumsum(b,axis=1)
```

### np.percentile()

- numpy.percentile()function used to compute the nth percentile of the given data (array elements) along the specified axis.

```python
a = np.array([100,200,300,400,500])
b = np.array([1,2,3,4,5])
a,b
```

### np.histogram()

- Numpy has a built-in numpy.histogram() function which represents the frequency of data distribution in the graphical form.

```python
c = np.random.randint(1,100,20)
c
```

```python
np.histogram(c,bins=[0,50,100])
```

```python
import matplotlib.pyplot as plt
plt.hist(c,bins=[0,50,100])
```

### np.corrcoef()

- Return Pearson product-moment correlation coefficients.

```python
salary = np.array([20000,40000,25000,35000,60000])
experience = np.array([1,3,2,4,2])

np.corrcoef(salary,experience)
```

### np.flip()

- The numpy.flip() function reverses the order of array elements along the specified axis, preserving the shape of the array.

```python
a = np.arange(20).reshape(4,5)
a
```

```python
np.flip(a,axis=0) # flip rows
```

### Set functions

- np.union1d
- np.intersect1d
- np.setdiff1d
- np.setxor1d
- np.in1d

```python
a = np.array([30,33,20,89,76,45,19,96])
b = np.array([30,33,20,100,150,200,19,300])
print(a,b)
```

```python
# np.union1d()

np.union1d(a,b)
```

```python
# np.intersect1d()

np.intersect1d(a,b)
```

```python
# np.setdiff1d()

np.setdiff1d(b,a)
```

```python
np.setxor1d(a,b)
```

```python
np.isin(a,33)
```
