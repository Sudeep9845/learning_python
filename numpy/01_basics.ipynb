### What is numpy?

NumPy is the fundamental package for scientific computing in Python. It is a Python library that provides a multidimensional array object, various derived objects (such as masked arrays and matrices), and an assortment of routines for fast operations on arrays, including mathematical, logical, shape manipulation, sorting, selecting, I/O, discrete Fourier transforms, basic linear algebra, basic statistical operations, random simulation and much more.


At the core of the NumPy package, is the ndarray object. This encapsulates n-dimensional arrays of homogeneous data types

### Numpy Arrays Vs Python Sequences

- NumPy arrays have a fixed size at creation, unlike Python lists (which can grow dynamically). Changing the size of an ndarray will create a new array and delete the original.

- The elements in a NumPy array are all required to be of the same data type, and thus will be the same size in memory.

- NumPy arrays facilitate advanced mathematical and other types of operations on large numbers of data. Typically, such operations are executed more efficiently and with less code than is possible using Python’s built-in sequences.

- A growing plethora of scientific and mathematical Python-based packages are using NumPy arrays; though these typically support Python-sequence input, they convert such input to NumPy arrays prior to processing, and they often output NumPy arrays.

### Numpy Array operations

```python
arr1 = np.arange(12).reshape(3,4)
arr2 = np.arange(12,24).reshape(3,4)
arr1
```

```python
# scalar operations

# airthmetic operations
# here arr1 is a 3x4 matrix and 2 is a scalar
print(arr1 * 2)
```

```python
# relational operations
print(arr1 > 5)
```

```python
# vector operations
# arithmetic operations
# here both arr1 and arr2 are 3x4 matrices
print(arr1 + arr2)
```

```python
# max and min
print(f"max of arr1: {np.max(arr1)},min of arr1:{np.min(arr1)}")
# row is 1 and col is 0
print(f"max of row1 : {np.max(arr1,axis=0)}")
print(f"max of col1 : {np.max(arr1,axis=1)}")
```

```python
# sum and prod

print(f"sum of arr1: {np.sum(arr1)}")
print(f"product of arr1: {np.prod(arr1)}")
```

```python
# mean 
arr4 = np.arange(1,13).reshape(3,4)
# mean_arr4 = np.mean(arr4,axis=1)
mean_arr4 = np.mean(arr4)
print(f"mean of arr4: {mean_arr4}")
```

```python
# median
median_arr4 = np.median(arr4)
print(f"median of arr4: {median_arr4}")
```

```python
# standard deviation and varianvce
variance_arr4 = np.var(arr4)
print(f"variance of arr4: {variance_arr4}")

# standard deviation
stddev_arr4 = np.std(arr4)
print(f"standard deviation of arr4: {stddev_arr4}")
```

```python
# trignometry functions (not used much)
print(np.sin(arr4))
```

```python
# dot products
m1 = np.arange(1,7).reshape(2,3)
m2 = np.arange(6,12).reshape(3,2)
dot_product = np.dot(m1,m2)
print(f"dot product of m1 and m2:\n{dot_product}")
```

```python
# log and exponential functions
print(f"Log of m1\n{np.log(m1)}\nExp of m1\n{np.exp(m1)}")
```

```python
# round/floor/ceil
random_array = np.random.random((2,3)) * 100
print(f"random array:\n{random_array}")
floored_arr = np.floor(random_array)
rounded_arr = np.round(random_array)
ceiled_arr = np.ceil(random_array)
print(f"floored array:\n{floored_arr}\nrounded array:\n{rounded_arr}\nceiled array:\n{ceiled_arr}")
```

### Stacking

```python
a4 = np.arange(16).reshape(4,4)
a5 = np.arange(16,32).reshape(4,4)
print(a4)
print(a5)
```

```python
# hstack
np.hstack((a4,a5))
```

```python
# vstack
np.vstack((a4,a5))
```
