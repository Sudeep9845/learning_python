## Pandas Dataframes

- A DataFrame is a 2-dimensional, size-mutable, and heterogeneous tabular data structure with labeled axes (rows and columns). It is primarily used for data manipulation, cleaning, and analysis in languages like Python
- Conceptually, a DataFrame is similar to an Excel spreadsheet or a SQL table, where each column represents a variable and each row represents an observation. 
- Single column is a Series
- Single row is called as "B-Series"

```python
# toss winner is match winner in percentage
toss_winner = ipl["TossWinner"]
match_winner = ipl["WinningTeam"]
toss_and_match_winners = ipl[toss_winner == match_winner]

percentage_of_toss_and_match_winners = (toss_and_match_winners.shape[0]/ipl.shape[0])*100
print(f"Toss winner is match winner in percentage: {percentage_of_toss_and_match_winners}")
```

#### To filter for movies that contain "Action" within a string that also contains other genres (e.g., "Action|Drama|War"), you should use the Pandas .str.contains() method instead of the equality operator (==).
  

```python
# action movies with rating higher than 7.5
action_movies = movies["genres"].str.contains("Action",na=False)
movies_above_7_5 = movies["imdb_rating"] > 7.5
movies[(action_movies) & (movies_above_7_5)][["title_x","genres","imdb_rating"]]
```

```python
# write a function that can return the track record of  2 teams against each other
def track_record(team1,team2):
    matches = ipl[((ipl["Team1"] == team1) & (ipl["Team2"] == team2)) | 
                 ((ipl["Team1"] == team2) & (ipl["Team2"] == team1))]
    if matches.empty:
        return f'No matches were played between {team1} and {team2}.'
    team1_wins = matches[matches["WinningTeam"] == team1].shape[0]
    team2_wins = matches[matches["WinningTeam"] == team2].shape[0]
    total_matches = matches.shape[0]

    result = {"total_matches":total_matches,
              "team1_wins":team1_wins, 
              "team2_wins":team2_wins}
    return result

def get_record(team1,team2):
    records = track_record(team1,team2)
    print(f"Total matches : {records["total_matches"]}\n{team1} Wins: {records["team1_wins"]}\n{team2} Wins: {records["team2_wins"]}")
    
team1 = "Royal Challengers Bangalore"
team2 = "Mumbai Indians"
get_record(team1,team2)
```
